### Task 10: Caught Speeding Logic

```python
def caught_speeding(speed, is_birthday):
    if is_birthday:
        speed -= 5

    if speed <= 60:
        return "No Ticket"
    elif 61 <= speed <= 80:
        return "Small Ticket"
    else:
        return "Big Ticket"

print("Q10 (Normal, 81 mph):", caught_speeding(81, False))
print("Q10 (Birthday, 81 mph):", caught_speeding(81, True))
```

### Bonus 1: Robust Dog Counter (Handling Punctuation)

```python
import string

def advanced_count_dog(s):
    words = s.lower().split()
    cleaned_words = [w.strip(string.punctuation) for w in words]
    return cleaned_words.count("dog")

print("Bonus 1:", advanced_count_dog("The dog, the dog! Watch the dog..."))
```

### Bonus 2: Palindrome checker with slicing

```python
def is_palindrome(word):
    return word.lower() == word[::-1].lower()

print("Bonus 2:", is_palindrome("Racecar"))
```

### Bonus 3: Find Topper in List of Dicts

```python
students = [
    {"name": "Zahra", "marks": 85},
    {"name": "Ali", "marks": 96},
    {"name": "Sara", "marks": 92},
    {"name": "Ahmed", "marks": 75},
    {"name": "Bilal", "marks": 88}
]

topper = max(students, key=lambda x: x["marks"])
print("Bonus 3: The topper is", topper["name"])
```
