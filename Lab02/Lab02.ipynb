## Task 7 · Climate of five cities (real data, PMD 1991–2020 normals)
Using `tmax` (mean daily maximum °C) and `rain` (mm), rows = `cities`, columns = `months`:
- **(a)** Shape of `tmax`.
- **(b)** Annual mean maximum temperature of each city.
- **(c)** Hottest month of each city (use `argmax` + `months`).
- **(d)** 5-city average for each month.
- **(e)** Number of months above 35 °C in each city.
- **(f)** Lahore's temperatures in °F (`F = C × 9/5 + 32`).

- **(h)** The wettest city in the monsoon, and what % of each city's yearly rain falls in the monsoon.

**Interpret:** in 1–2 lines, which result surprised you, and why?

```python
# T7: your code here
print("T7a (Shape of tmax):", tmax.shape)

annual_mean_tmax = tmax.mean(axis=1)
print("T7b (Mean Tmax °C):", np.round(annual_mean_tmax, 2))

hottest_month_indices = tmax.argmax(axis=1)
hottest_months = months[hottest_month_indices]
print("T7c (Hottest Month per City):", list(zip(cities, hottest_months)))

five_city_monthly_avg = tmax.mean(axis=0)
print("T7d (5-City Monthly Avg °C):", np.round(five_city_monthly_avg, 2))

months_above_35 = (tmax > 35.0).sum(axis=1)
print("T7e (Months > 35°C):", list(zip(cities, months_above_35)))

lahore_f = tmax[0] * 9 / 5 + 32
print("T7f (Lahore Tmax °F):", np.round(lahore_f, 1))

monsoon_rain = rain[:, 6:9].sum(axis=1)
print("T7g (Monsoon Rain mm):", list(zip(cities, np.round(monsoon_rain, 1))))

yearly_rain = rain.sum(axis=1)
wettest_monsoon_city = cities[monsoon_rain.argmax()]
monsoon_share_pct = (monsoon_rain / yearly_rain) * 100
print(f"T7h: Wettest monsoon city is {wettest_monsoon_city} ({monsoon_rain.max():.1f} mm)")
print("T7h (Monsoon Share of Annual Rain %):", np.round(monsoon_share_pct, 1))
```

Interpretation: Islamabad receives almost 70% more monsoon rainfall than Lahore (763.0 mm vs 454.3 mm). In contrast, Quetta receives almost no monsoon rain (<18 mm), getting the majority of its precipitation during the winter season.

## Task 9 · Grace marks
Using `marks`:
- **(a)** Add 5 grace marks but cap at 100 (`np.clip`).
- **(b)** Pass % (≥ 50) after grace.
- **(c)** Index of students who passed **only because of** the grace marks.
- **(d)** z-score of each student after grace.

```python
# T9: your code here
marks_after_grace = np.clip(marks + 5, 0, 100)
print("T9a (Marks after grace):", marks_after_grace)

pass_pct = (marks_after_grace >= 50).mean() * 100
print(f"T9b (Pass %): {pass_pct:.1f}%")

passed_only_grace = np.where((marks < 50) & (marks_after_grace >= 50))[0]
print("T9c (Indices saved by grace):", passed_only_grace)

mean_grace = marks_after_grace.mean()
std_grace = marks_after_grace.std()
z_scores = (marks_after_grace - mean_grace) / std_grace
print("T9d (Z-scores):", np.round(z_scores, 2))
```
