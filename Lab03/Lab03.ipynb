## Task 3 · Clean the data: the six types of problem
Fix the problems **in this order** (text first, then duplicates):
- **(a) Missing values:** count them per column. Drop ads with no `Brand` or no `Model`; fill missing `Registered City`, `Fuel` and `Transaction Type` with `"Unknown"`; leave `Year` and `KMs Driven` as NaN. Print how many rows were removed.
- **(b) Formatting:** remove extra spaces and fix the case of city names (`.str.strip().str.title()`); remove `"Rs"` and commas from `Price`.
- **(c) Data types:** convert `Price` to `int64` and `Year` to `Int64`.
- **(d) Inconsistent categories:** `"Sukkar"` → `"Sukkur"` and `"Corrolla Altis"` → `"Corolla Altis"`. Print the number of different cities before and after.
- **(e) Duplicates:** count exact duplicate ads, drop them, print the new row count.
- **(f) Invalid values:** set `KMs Driven` to NaN when it is below 100, above 1,000,000, or the placeholder 12345 / 123456. Drop ads with the fake prices 12,345,678 or 87,654,321, and Suzuki/Daihatsu ads above 5,000,000.
- **(g) Self-check:** your final shape must be **(20484, 9)**.

```python

# T3: your code here
# (a) Missing values
print("T3a missing values per column:\n", df.isna().sum())
initial_count = len(df)
df = df.dropna(subset=["Brand", "Model"])
rows_dropped_missing = initial_count - len(df)
print("\nT3a rows removed without Brand or Model:", rows_dropped_missing)

df["Registered City"] = df["Registered City"].fillna("Unknown")
df["Fuel"] = df["Fuel"].fillna("Unknown")
df["Transaction Type"] = df["Transaction Type"].fillna("Unknown")

# (b) Formatting: strip spaces, Title Case, clean Price
cities_before = df["Registered City"].nunique()
df["Registered City"] = df["Registered City"].astype(str).str.strip().str.title()
df["Price"] = df["Price"].astype(str).str.replace("Rs", "", regex=False).str.replace(",", "", regex=False).str.strip()

# (c) Data types: Price to int64, Year to Int64
df["Price"] = df["Price"].astype("int64")
df["Year"] = df["Year"].astype("Int64")
print("\nT3c dtypes:\n", df[["Price", "Year"]].dtypes)

# (d) Inconsistent categories
df["Registered City"] = df["Registered City"].replace({"Sukkar": "Sukkur"})
df["Model"] = df["Model"].replace({"Corrolla Altis": "Corolla Altis"})
cities_after = df["Registered City"].nunique()
print(f"\nT3d distinct cities: {cities_before} before -> {cities_after} after")

# (e) Duplicates
dup_count = df.duplicated().sum()
print("\nT3e exact duplicate ads:", dup_count)
df = df.drop_duplicates()
print("T3e new row count after dropping duplicates:", len(df))

# (f) Invalid values
bad_kms_mask = (df["KMs Driven"] < 100) | (df["KMs Driven"] > 1_000_000) | (df["KMs Driven"].isin([12345, 123456]))
df.loc[bad_kms_mask, "KMs Driven"] = np.nan

fake_prices = df["Price"].isin([12345678, 87654321])
bad_budget_cars = (df["Brand"].isin(["Suzuki", "Daihatsu"])) & (df["Price"] > 5_000_000)
df = df[~(fake_prices | bad_budget_cars)]

# (g) Self-check
print("\nT3g final shape:", df.shape)
assert df.shape == (20484, 9), f"Self-check failed: expected (20484, 9), got {df.shape}"
print("T3g self-check passed: Shape is exactly (20484, 9)")
```

## Task 4 · Analyse
- **(a)** New column `Price_lakh` = price in lakh rupees (÷ 100,000).
- **(b)** New column `Age` = 2020 − `Year`.
- **(c)** The 5 most common brands.
- **(d)** The 5 cities with the most ads.
- **(e)** Median `Price_lakh` for each fuel type, highest first.
- **(f)** Median `Price_lakh` of a **Corolla GLI** for each year 2010–2018. How much value does it lose per year?
- **(g)** The 5 most expensive ads. Does one of them look wrong? Why?
- **(h)** Number of ads and median price for Cash vs Installment/Leasing. Interpret in 1–2 lines.

```python

# T4: your code here
# (a) New column Price_lakh
df["Price_lakh"] = df["Price"] / 100_000

# (b) New column Age
df["Age"] = 2020 - df["Year"]

# (c) 5 most common brands
print("T4c top 5 brands:\n", df["Brand"].value_counts().head(5))

# (d) 5 cities with the most ads
print("\nT4d top 5 cities:\n", df["Registered City"].value_counts().head(5))

# (e) Median Price_lakh per fuel type
fuel_medians = df.groupby("Fuel")["Price_lakh"].median().sort_values(ascending=False)
print("\nT4e median price per fuel type (in lakh):\n", fuel_medians)

# (f) Median Price_lakh of Corolla GLI (2010–2018)
gli_ads = df[(df["Model"] == "Corolla GLI") & (df["Year"] >= 2010) & (df["Year"] <= 2018)]
gli_by_year = gli_ads.groupby("Year")["Price_lakh"].median()
print("\nT4f Corolla GLI median price by year (2010-2018):\n", gli_by_year)
annual_loss = (gli_by_year.loc[2018] - gli_by_year.loc[2010]) / (2018 - 2010)
print(f"T4f estimated value loss/depreciation: ~{annual_loss:.2f} lakh/year")

# (g) 5 most expensive ads
expensive_cars = df.nlargest(5, "Price")[["Brand", "Model", "Year", "Price_lakh"]]
print("\nT4g 5 most expensive ads:\n", expensive_cars)

# (h) Cash vs Installment/Leasing counts and medians
transaction_stats = df.groupby("Transaction Type")["Price_lakh"].agg(["count", "median"])
print("\nT4h Cash vs Installment/Leasing:\n", transaction_stats)
```

**Task 4 Interpretations:**
* **4 (f) Corolla GLI Depreciation:** Between 2010 (median 12.25 lakh) and 2018 (median 23.20 lakh), a Corolla GLI gains roughly ~1.37 lakh per newer model year, showing an average depreciation of ~1.37 lakh per year of vehicle age.
* **4 (g) Suspicious Ad Analysis:** The **2015 Toyota Prius** listed at **357.5 lakh (Rs 3.575 Crore)** is an erroneous entry/typo. While Land Cruisers legitimately trade at several crores, a 2015 Prius typically commands 25–35 lakh in Pakistan.
* **4 (h) Cash vs Installment Interpretation:** Over 95% of ads (`19,634`) are listed under Cash terms with a median price of `6.9 lakh`, compared to only `850` ads under Installment/Leasing. The much lower installment median price (`3.0 lakh`) occurs because sellers list the initial down-payment/advance required rather than total amortized price.
